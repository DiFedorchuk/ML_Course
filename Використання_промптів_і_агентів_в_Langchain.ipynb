````python
import json
import os

with open('creds.json') as file:
  creds = json.load(file)
os.environ["HF_TOKEN"] = creds['HUGGINGFACEHUB_API_TOKEN']
````

### Завдання 2: Створення параметризованого промпта для генерації тексту
Тепер ми хочемо оновити попередній фукнціонал так, аби в промпт ми могли передавати тему як параметр. Для цього скористайтесь `PromptTemplate` з `langchain` і реалізуйте параметризований промпт та виклик моделі з ним.

Запустіть оновлений функціонал (промпт + модел) для пояснень про теми
- "Баєсівські методи в машинному навчанні"
- "Трансформери в машинному навчанні"
- "Explainable AI"

Виведіть результати відпрацювання моделі на екран.

````python
from langchain_core.prompts import PromptTemplate
prompt_template = PromptTemplate(
    input_variables=["topic"],
    template="Що таке {topic}? Відповідь до 200 символів, коротка."
)
topics = [
    "Баєсівські методи в машинному навчанні",
    "Трансформери в машинному навчанні",
    "Explainable AI"
]
print("Генерація пояснень для наступних тем:")
for topic in topics:
    formatted_prompt = prompt_template.format(topic=topic)
    print(f"\nТема: {topic}")
    print(f"Промпт: {formatted_prompt}")
    response = chat.invoke(formatted_prompt)
    print(f"Пояснення: {response.content}")
````

Для взаємодії з пошуком там необхідно створити `Tool`. В лекції ми використовували `serpapi`. Можна продовжити користуватись ним, або обрати інше АРІ для пошуку (вони в тому числі є безкоштовні).
````python
import uuid
import builtins
builtins.uuid = uuid
from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace
from langchain_community.tools import DuckDuckGoSearchRun
from langchain_experimental.utilities import PythonREPL
from langchain_core.tools import Tool
from langgraph.prebuilt import create_react_agent

with open("creds.json") as file:
    creds = json.load(file)

os.environ["HF_TOKEN"] = creds["HUGGINGFACEHUB_API_TOKEN"]
base_llm = HuggingFaceEndpoint(
    repo_id="Qwen/Qwen2.5-7B-Instruct",
    max_new_tokens=1024,
    top_k=10,
    top_p=0.95,
    temperature=0.01,
    repetition_penalty=1.03,
    huggingfacehub_api_token=os.environ["HF_TOKEN"],
)

llm = ChatHuggingFace(llm=base_llm)
python_tool = Tool(
    name="python_interpreter",
    func=PythonREPL().run,
    description=(
        "Use for calculations, data processing, or logic. "
        "Input must be valid Python code and print the result."
    ),
)
search_tool = Tool(
    name="DuckDuckGo_Search",
    func=DuckDuckGoSearchRun().run,
    description="Search the internet for recent and factual information.",
)

tools = [search_tool, python_tool]
agent = create_react_agent(llm, tools)
user_query = (
    "Знайди 5 останніх наукових публікацій на тему 'штучний інтелект'. "
    "Для кожної публікації вкажи: назву публікації, авторів та короткий опис."
)

print("Executing agent...\n")
result = agent.invoke(
    {"messages": [("human", user_query)]}
)
print("\nFinal Answer:\n")
print(result["messages"][-1].content)
````
